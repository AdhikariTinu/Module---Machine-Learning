# Regression

### Q.1. What is Simple Linear Regression?

--> Simple Linear Regression (SLR) is a statistical and machine learning technique used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line through the data points.

The mathematical equation is:

Y = mX + c

Where:

Y is the dependent variable

X is the independent variable

m is the slope

c is the intercept

Example: If we want to predict a student’s marks (Y) based on hours studied (X), simple linear regression can be used to find the best-fit line that explains how marks increase with study hours.

### Q.2. What are the key assumptions of Simple Linear Regression?

--> Linearity: The relationship between X and Y must be linear.

Independence: Observations must be independent of each other.

Homoscedasticity: Constant variance of errors across all values of X.

Normality of errors: Residuals should be normally distributed.

No significant outliers: Extreme values should not heavily influence the model.

### Q.3. What does the coefficient m represent in the equation Y=mX+c?

--> The coefficient m represents the slope of the line.

### Q.4. What does the intercept c represent in the equation Y = mX + c?

--> The intercept c represents the value of Y when X = 0.

### Q.5. How do we calculate the slope m in Simple Linear Regression?

--> The slope is calculated using the formula:

m = Σ(X − X̄)(Y − Ȳ) / Σ(X − X̄)²

### Q.6. What is the purpose of the least squares method?

--> The least squares method finds the best-fitting line by minimizing the sum of squared residuals, ensuring the most accurate predictions possible.

### Q.7. How is the coefficient of determination (R²) interpreted?

--> R² indicates the proportion of variance in Y explained by X.

For example, R² = 0.80 means 80% of variability in Y is explained by the model.

### Q.8. What is Multiple Linear Regression?

--> Multiple Linear Regression models the relationship between one dependent variable and two or more independent variables.

General form:

Y = b₀ + b₁X₁ + b₂X₂ + … + bₙXₙ

### Q.9. What is the main difference between Simple and Multiple Linear Regression?

--> Simple: One independent variable

--> Multiple: Two or more independent variables

### Q.10. What are the key assumptions of Multiple Linear Regression?

--> Linearity

Independence

Homoscedasticity

Normality of residuals

No multicollinearity

### Q.11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

--> Heteroscedasticity occurs when residual variance is not constant.

Impact:

Incorrect standard errors

Unreliable hypothesis testing

Example: Higher prediction errors for expensive houses compared to cheaper ones.

### Q.12. How can you improve a Multiple Linear Regression model with high multicollinearity?

--> Remove correlated features

Use VIF

Apply PCA

Use Ridge or Lasso regression

### Q.13. What are some common techniques for transforming categorical variables for use in regression models?

--> One-hot encoding

Label encoding

Dummy variables

Example: City => Delhi, Mumbai, Chennai becomes binary columns.

### Q.14. What is the role of interaction terms in Multiple Linear Regression?

--> Interaction terms show how one variable’s effect depends on another.

Example: Effect of advertising budget depends on season.

### Q.15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

--> Simple: Y when X = 0

Multiple: Y when all X variables are zero

### Q.16. What is the significance of the slope in regression analysis, and how does it affect predictions?

--> The slope in regression analysis is the engine of the model. It tells the story of change, direction, and intensity, and it directly drives every prediction the model makes.

Y = mX + c

The sign of the slope tells you which way the relationship leans:

Positive slope (m > 0):
As X increases, Y increases.
Example: More hours studied → higher marks.

Negative slope (m < 0):
As X increases, Y decreases.
Example: More speed → lower fuel efficiency.

Zero slope (m = 0):
No relationship between X and Y.

### Q.17. How does the intercept in a regression model provide context for the relationship between variables?

--> The intercept provides a baseline value of Y and gives meaning to changes introduced by the slope.

### Q.18. What are the limitations of using R² as a sole measure of model performance?

--> Does not imply causation => A high R² only means the model explains variance, not that the independent variables cause changes in the dependent variable.

Increases with added predictors => R² never decreases when new variables are added, even if those variables are irrelevant.

Cannot detect overfitting => A model can have a very high R² on training data and still fail badly on new, unseen data.

### Q.19. How would you interpret a large standard error for a regression coefficient?

--> Low Precision of the Coefficient Estimate

A large standard error means the estimated coefficient is not precise. The true value of the coefficient could lie across a wide range.


--> Weak Statistical Significance

Standard error directly affects the t-statistic:

t = Coefficient / Standard Error

If the standard error is large:

The t-value becomes small

The p-value increases

This often leads to the coefficient being statistically insignificant, even if its estimated value is large.

--> Possible Causes of a Large Standard Error

Multicollinearity: Predictor is highly correlated with other variables

Small sample size: Not enough data to estimate the effect reliably

High variability in data: Noisy or inconsistent observations

Model misspecification: Important variables are missing or wrongly included.

--> Impact on Predictions

While the model may still make predictions, a coefficient with a large standard error contributes unstable and unreliable effects, especially when extrapolating.


--> Example

Suppose a salary prediction model gives:

Experience coefficient = 4,000
Standard error = 3,500

This means the true effect of experience could plausibly be very small or very large, making conclusions uncertain.

### Q.20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

--> Heteroscedasticity appears as funnel or cone-shaped patterns in residual plots

It causes incorrect standard errors and hypothesis tests

Addressing it ensures valid inference and reliable predictions

### Q.21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?

--> High R² but low adjusted R² indicates unnecessary or irrelevant variables

The model is likely overfitting

Adjusted R² provides a more realistic measure of model quality in multiple regression

### Q.22. Why is it important to scale variables in Multiple Linear Regression?

--> Scaling ensures fair coefficient comparison

Improves numerical stability and convergence

Essential for Ridge and Lasso regression

Prevents dominance of large-scale variables

### Q.23. What is polynomial regression?

--> Polynomial regression models the relationship between an independent variable X and a dependent variable Y by including higher-degree powers of X (such as X², X³, etc.).

For a degree-n polynomial:

Y = b₀ + b₁X + b₂X² + b₃X³ + … + bₙXⁿ

Where:

Y = dependent variable

X = independent variable

b₀, b₁, …, bₙ = regression coefficients

### Q.24. How does polynomial regression differ from linear regression?

--> Linear regression fits a straight line

Polynomial regression fits a curve

Polynomial regression handles non-linear data

Higher-degree polynomials increase flexibility but risk overfitting

### Q.25. When is polynomial regression used?

--> Used when data shows a curved, non-linear trend

Applied when linear regression underfits

Suitable for smooth and continuous relationships

Degree must be chosen carefully to avoid overfitting

### Q.26. What is the general equation for polynomial regression?

--> For a polynomial of degree n, the equation is:

Y = b₀ + b₁X + b₂X² + b₃X³ + … + bₙXⁿ

Where:

Y = dependent variable

X = independent variable

b₀ = intercept

b₁, b₂, …, bₙ = regression coefficients

n = degree of the polynomial

### Q.27. Can polynomial regression be applied to multiple variables?

--> Yes. Polynomial regression can be applied to multiple variables, called multivariate polynomial regression. It includes powers and interaction terms of the independent variables.

It’s implemented in Python using PolynomialFeatures from scikit-learn. Overfitting can occur with high degrees or many variables.

### Q.28. What are the limitations of polynomial regression?

--> Overfitting – High-degree polynomials can fit the training data too closely, capturing noise rather than the true pattern.

Poor extrapolation – Polynomials can behave unpredictably outside the range of training data.

Complexity grows quickly – With multiple variables and higher degrees, the number of terms increases rapidly, making the model hard to interpret.

Sensitive to outliers – Large deviations in data can drastically affect the fitted curve.

Multicollinearity – Polynomial terms can be highly correlated with each other, affecting coefficient stability.

Not always interpretable – Higher-degree polynomials are mathematically accurate but difficult to explain meaningfully.

### Q.29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

--> R² (Coefficient of Determination) – Measures how much variance in the dependent variable is explained by the model. Higher is better, but it always increases with degree, so it can be misleading.

Adjusted R² – Adjusts for the number of predictors. Useful because it penalizes unnecessary complexity.

Mean Squared Error (MSE) / Root Mean Squared Error (RMSE) – Measures average prediction error. Lower values indicate a better fit.

Cross-Validation (e.g., k-fold CV) – Splits data into training and validation sets to evaluate generalization performance. Prevents overfitting.

AIC / BIC (Akaike / Bayesian Information Criteria) – Penalize model complexity to find the simplest model that fits well. Lower values are preferred.

Residual Analysis – Plot residuals to check for patterns. Residuals should be randomly distributed; systematic patterns indicate poor fit or underfitting/overfitting.

Rule of thumb: Use cross-validation + adjusted R² or RMSE to pick the degree that generalizes best, rather than just the one that fits the training data perfectly.
