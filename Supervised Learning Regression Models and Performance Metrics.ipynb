### Q.1. What is Simple Linear Regression (SLR)? Explain its purpose.

--> Simple Linear Regression (SLR) is a statistical and machine learning technique used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line to the observed data.

The main idea of SLR is to find how changes in the input variable affect the output variable.

Purpose of Simple Linear Regression

To understand the relationship between two variables

To predict future values of the dependent variable

To quantify the effect of one variable on another

To identify trends and patterns in data

For example, predicting salary based on years of experience.

### Q.2. What are the key assumptions of Simple Linear Regression?

--> Simple Linear Regression works correctly only when certain assumptions are satisfied:

Linearity >> The relationship between X and Y should be linear.

Independence >> Observations should be independent of each other.

Homoscedasticity >> The variance of errors should remain constant for all values of X.

Normality of Errors >> The residuals (errors) should follow a normal distribution.

No Multicollinearity >> Since SLR has only one independent variable, this is naturally satisfied.

If these assumptions are violated, predictions and interpretations may be inaccurate.

### Q.3. Write the mathematical equation for a simple linear regression model and explain each term.

--> The mathematical equation of Simple Linear Regression is:

y = mx + c

Explanation of each term:

y -> Dependent variable (output)

x -> Independent variable (input)

c (Intercept) -> Value of y when x = 0

m (Slope) -> Change in y for a one-unit change in x

This equation represents a straight line that best fits the data.

### Q.4. Provide a real-world example where simple linear regression can be applied.

--> A real-world example of Simple Linear Regression is:

Predicting House Prices

Independent Variable (X): Size of the house (in square feet)

Dependent Variable (Y): House price

Using SLR, we can estimate how much the house price increases when the size increases by one unit.

### Q.5. What is the method of least squares in linear regression?

--> The method of least squares is a technique used to find the best-fitting line in linear regression.

How it works:

It minimizes the sum of squared differences between actual values and predicted values.

These differences are called residuals.

Mathematically, it minimizes:

=> ∑ ( y_actual - y_predicted ) ²

Why squared errors?

Penalizes large errors more

Ensures a unique optimal solution

This method helps in finding the most accurate values of slope and intercept.

### Q.6. What is Logistic Regression? How does it differ from Linear Regression?

--> Logistic Regression is a classification algorithm used when the dependent variable is categorical, usually binary (0 or 1).

It predicts the probability of an event occurring using a sigmoid function.

Differences between Linear and Logistic Regression

| Feature      | Linear Regression | Logistic Regression |
| ------------ | ----------------- | ------------------- |
| Output       | Continuous        | Categorical         |
| Use case     | Prediction        | Classification      |
| Function     | Straight line     | Sigmoid curve       |
| Output range | (-∞, +∞)          | (0, 1)              |
| Example      | Predict salary    | Predict pass/fail   |


### Q.7. Name and briefly describe three common evaluation metrics for regression models.

> Mean Absolute Error (MAE):

>> Average of absolute differences between actual and predicted values.

>> Easy to interpret.

> Mean Squared Error (MSE): 

>> Average of squared differences between actual and predicted values.

>> Penalizes large errors more.

> Root Mean Squared Error (RMSE):

>> Square root of MSE.

>> Expressed in the same unit as the target variable.

### Q.8. What is the purpose of the R-squared metric in regression analysis?

--> R-squared (R²) measures how well the regression model explains the variability in the dependent variable.

Key points:

R² ranges from 0 to 1

Higher R² means better model fit

R² = 0 >> Model explains nothing

R² = 1 >> Perfect prediction

Purpose:

To evaluate model performance

To compare multiple regression models

To understand explanatory power

### Q.9. Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

import warnings
warnings.filterwarnings('ignore')
```

```python
data = {
    'x': [1,2,3,4,5,6,7,8,9],
    'y': [2,4,6,8,10,12,14,16,18]
}
df = pd.DataFrame(data)

sns.scatterplot(data=df, x='x', y='y', marker='o', color='black',label= 'Actual Data Points')
plt.title('Scatterplot y v/s x without best-fit line')
plt.grid()
plt.show()

X = df[['x']]
y = df['y']

model = LinearRegression()
model.fit(X,y)

sns.scatterplot(data=df, x='x', y='y', marker='o', color='black', label= 'Actual Data Points')
plt.plot(df['x'], model.predict(df[['x']]), 'r--', label='Predicted Line')
plt.title('Scatterplot y v/s x with best-fit line')
plt.legend()
plt.grid()
plt.show()

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

### Q.10. How do you interpret the coefficients in a simple linear regression model?

--> In Simple Linear Regression:

Slope (m):
Indicates how much the dependent variable changes when the independent variable increases by one unit.

Intercept (c):
Represents the value of the dependent variable when the independent variable is zero.

Example Interpretation:

If the slope is 5, then for every 1-unit increase in X, Y increases by 5 units.

This helps understand the strength and direction of the relationship.
